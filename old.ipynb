````python
# 📅 Date columns processing
print("📅 Processing date columns...")

# Convert date columns to datetime
date_columns = ['order_date', 'ship_date']
for col in date_columns:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(df_clean[col])
        print(f"✅ Converted {col} to datetime")

print(f"📊 Date range: {df_clean['order_date'].min()} to {df_clean['order_date'].max()}")
````

````python
# 🔢 Numerical data validation and cleaning
print("🔢 Validating numerical columns...")

numerical_cols = ['discount', 'profit', 'quantity', 'sales', 'profit_margin']

for col in numerical_cols:
    if col in df_clean.columns:
        # Check for negative values where they shouldn't exist
        if col in ['quantity', 'sales']:
            negative_count = (df_clean[col] < 0).sum()
            if negative_count > 0:
                print(f"⚠️  Found {negative_count} negative values in {col}")
        
        # Check for extreme outliers (beyond 3 standard deviations)
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        outliers = ((df_clean[col] < (Q1 - 3 * IQR)) | (df_clean[col] > (Q3 + 3 * IQR))).sum()
        print(f"📊 {col}: {outliers} extreme outliers detected")

print("✅ Numerical validation completed")
````

````python
# 📊 Cleaned dataset overview
print("📊 Cleaned Dataset Overview:")
print("="*50)
print(f"📋 Shape: {df_clean.shape}")
print(f"🔢 Numerical columns: {len(df_clean.select_dtypes(include=[np.number]).columns)}")
print(f"🔤 Categorical columns: {len(df_clean.select_dtypes(include=['object']).columns)}")
print(f"📅 Date columns: {len(df_clean.select_dtypes(include=['datetime64']).columns)}")

# Display cleaned data sample
df_clean.head()
````

````python
# 🔄 Data Transformation and Feature Creation

print("🔄 Starting Data Transformation...")
print("="*50)

# Create the target variable: Profit Binary (1 = Profitable, 0 = Unprofitable)
df_clean['profit_binary'] = (df_clean['profit'] > 0).astype(int)
profit_distribution = df_clean['profit_binary'].value_counts()
print(f"💰 Profit Distribution:")
print(f"   Profitable transactions: {profit_distribution[1]} ({profit_distribution[1]/len(df_clean)*100:.1f}%)")
print(f"   Unprofitable transactions: {profit_distribution[0]} ({profit_distribution[0]/len(df_clean)*100:.1f}%)")

print("✅ Target variable 'profit_binary' created successfully")
````

````python
# 💰 Financial metrics and derived features
print("💰 Creating financial metrics...")

# Calculate revenue (sales - profit)
df_clean['revenue'] = df_clean['sales'] - df_clean['profit']

# Revenue per unit
df_clean['revenue_per_unit'] = df_clean['revenue'] / df_clean['quantity']

# Profit per unit  
df_clean['profit_per_unit'] = df_clean['profit'] / df_clean['quantity']

# Discount amount in dollars
df_clean['discount_amount'] = df_clean['sales'] * df_clean['discount']

# Sales without discount
df_clean['sales_without_discount'] = df_clean['sales'] / (1 - df_clean['discount'])

print("✅ Financial metrics created:")
print("   - revenue, revenue_per_unit, profit_per_unit")
print("   - discount_amount, sales_without_discount")

# Display sample with new features
df_clean[['sales', 'profit', 'revenue', 'profit_margin', 'profit_binary']].head()
````

````python
# 🏷️ Categorical Feature Engineering
print("🏷️ Creating categorical features...")

# Create discount categories
df_features['discount_category'] = pd.cut(df_features['discount'], 
                                        bins=[-0.01, 0, 0.2, 0.5, 1.0], 
                                        labels=['No_Discount', 'Low', 'Medium', 'High'])

# Create sales volume categories
sales_quartiles = df_features['sales'].quantile([0.25, 0.5, 0.75]).values
df_features['sales_category'] = pd.cut(df_features['sales'], 
                                     bins=[-np.inf] + list(sales_quartiles) + [np.inf], 
                                     labels=['Low_Sales', 'Medium_Sales', 'High_Sales', 'Very_High_Sales'])

# Create profit categories
df_features['profit_category'] = pd.cut(df_features['profit'], 
                                      bins=[-np.inf, 0, 50, 200, np.inf], 
                                      labels=['Loss', 'Low_Profit', 'Medium_Profit', 'High_Profit'])

print("✅ Categorical features created: discount_category, sales_category, profit_category")

# Display sample of new features
sample_features = ['discount_category', 'sales_category', 'profit_category', 'season']
df_features[sample_features].head()
````

````python
# 📊 Aggregated Features - Customer Level
print("📊 Creating customer-level aggregated features...")

# Customer aggregations
customer_agg = df_features.groupby('customer').agg({
    'sales': ['sum', 'mean', 'count'],
    'profit': ['sum', 'mean'],
    'profit_binary': 'mean',
    'order_id': 'nunique'
}).round(2)

# Flatten column names
customer_agg.columns = ['_'.join(col).strip() for col in customer_agg.columns]
customer_agg = customer_agg.rename(columns={'order_id_nunique': 'total_orders'})

# Add customer performance metrics
customer_agg['avg_order_value'] = customer_agg['sales_sum'] / customer_agg['total_orders']
customer_agg['customer_profitability_rate'] = customer_agg['profit_binary_mean']

print(f"✅ Customer aggregations created for {len(customer_agg)} unique customers")
print("   Features: total_sales, avg_sales, order_count, total_profit, avg_profit, profitability_rate")

# Display customer aggregation sample
customer_agg.head()
````

````python
# 🗂️ Product and Regional Aggregations
print("🗂️ Creating product and regional aggregated features...")

# Product category aggregations
product_agg = df_features.groupby(['category', 'subcategory']).agg({
    'profit_binary': 'mean',
    'profit_margin': 'mean',
    'sales': 'mean'
}).round(3)

product_agg.columns = ['category_subcategory_profit_rate', 'category_subcategory_avg_margin', 'category_subcategory_avg_sales']

# Regional aggregations
regional_agg = df_features.groupby('region').agg({
    'profit_binary': 'mean',
    'profit_margin': 'mean',
    'sales': 'mean'
}).round(3)

regional_agg.columns = ['region_profit_rate', 'region_avg_margin', 'region_avg_sales']

print("✅ Product and regional aggregations created")
print("   Product features: category_subcategory profit metrics") 
print("   Regional features: region-level performance metrics")

# Display samples
print("\n📋 Product Category Performance Sample:")
product_agg.head()
````
